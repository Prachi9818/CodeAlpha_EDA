```python
mean_ = df['Price'].mean()
print("mean of Price column:",mean_)
```

```python
df['Price'] = df['Price'].fillna(mean_)
```

```python
median_ = df['Price'].median()
print("median of Price column:",median_)
```

```python
df.isnull().sum()
```

```python
df= df.dropna(subset=['Company'])
df= df.dropna(subset=['TypeName'])
df= df.dropna(subset=['Inches'])
df= df.dropna(subset=['ScreenResolution'])
df= df.dropna(subset=['Cpu'])
df= df.dropna(subset=['Ram'])
df= df.dropna(subset=['Memory'])
df= df.dropna(subset=['Gpu'])
df= df.dropna(subset=['OpSys'])
df= df.dropna(subset=['Weight'])
df= df.dropna(subset=['Price'])
df.isnull().sum()
```

```python
df['Price'] = df['Price'].fillna(df['Price'].mean())
df.isnull().sum()
```

```python
#Filtering
filtered_df = df[df['Company'] == 'Apple']
filtered_df
```

```python
#Filtering
filtered_df = df[df['Company'] == 'Samsung']
filtered_df
```

```python
#Filtering
filtered_df = df[df['Price'] > 50000]
filtered_df
```

```python
#Sorting
sorted_df = filtered_df.sort_values(by='Weight', ascending=False)
sorted_df
```

```python
#Sorting 
sorted_df = filtered_df.sort_values(by='Price', ascending=False)
sorted_df
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#UNIVARIATE ANALYSIS :- Price - histogram
sns.histplot(df['Price'], kde= True)
plt.title("Distribution of Laptop Prices")
```

```python
#UNIVARIATE ANALYSIS :- Company - countplot
sns.countplot(x='Company', data=df)
plt.xticks(rotation=90)
plt.title("Distribution of Laptop Companies")
```

```python
#UNIVARIATE ANALYSIS :- Ram - countplot
sns.countplot(x='Ram', data=df)
plt.title("Distribution of Laptop RAM")
plt.show()
```

```python
#UNIVARIATE ANALYSIS :- Weight - Boxplot
plt.figure(figsize=(12,6.8))
sns.boxplot(x=df['Weight'])
plt.title("Distribution of Laptop Weights")
plt.show()
```

```python
#BIVARIATE ANALYSIS :- Price vs Ram - scatterplot
plt.figure(figsize=(12,6.8))
sns.scatterplot(x='Ram', y='Price', data=df)
plt.title("Price vs RAM")
plt.show()
```

```python
#BIVARIATE ANALYSIS :- Price vs Company - Boxplot
plt.figure(figsize=(12,6.8))
sns.boxplot(x='Company', y='Price', data=df)
plt.title("Price vs Company")
plt.xticks(rotation=90)
plt.show()
```

```python
#BIVARIATE ANALYSIS :- Weight vs Price - Scatterplot
plt.figure(figsize=(12,6.8))
sns.scatterplot(x='Weight', y='Price', data=df)
plt.title("Weight vs Price")
plt.xticks(rotation=90)
plt.show()
```

```python
#MULTIVARIATE ANALYSIS :- Heatmap
plt.figure(figsize=(12,6.8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.show()
```
